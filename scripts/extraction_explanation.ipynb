## Data Extraction

The air-quality data used in this project was obtained from Excel workbooks covering two different data formats: the historical **2016–2020 dataset** and the more recent **2024–2025 datasets**. Because these sources use different spreadsheet structures, the extraction process was handled separately for each period.

The **2024 and 2025 workbooks** contain a dedicated `Ratnapark` worksheet where each row represents a single day and the columns directly contain the date and pollutant measurements (`PM2.5`, `PM10`, and `TSP`). Therefore, these sheets could be extracted directly using `pandas.read_excel()` and saved as CSV files without changing the original measurements or handling missing values.

The **2016–2020 workbook**, however, uses a different structure. The `Ratnapark` worksheet contains four separate pollutant sections: `PM1`, `PM2.5`, `PM10`, and `TSP`. Within each section, the data for the five years (2016–2020) are arranged horizontally as pairs of `Date` and `Value` columns. Therefore, the extraction process first identified each pollutant section, separated the yearly date/value pairs, and then reshaped them into a single chronological table. The individual pollutant datasets were subsequently merged using the date as the common key.

No missing-value imputation, interpolation, outlier removal, or data filtering based on pollutant availability was performed during extraction. Missing measurements from the original Excel files were intentionally preserved so that the extent and pattern of missing data can be analyzed separately during the data-cleaning and exploratory-data-analysis stages.
